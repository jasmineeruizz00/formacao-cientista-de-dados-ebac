**Efetividade de Duas Estratégias de Ensino**

Imagine que uma escola esteja avaliando a eficácia de duas estratégias de ensino de matemática para alunos do ensino médio. Eles querem determinar se há uma diferença significativa no desempenho médio dos alunos entre as duas estratégias.

# **Hipóteses:**

* Hipótese nula (H0): A média das notas dos alunos na estratégia A é igual à média das notas dos alunos na estratégia B.
* Hipótese alternativa (H1): A média das notas na Estratégia B é maior do que a média das notas na Estratégia A.

# **Dados:**

* Amostra da Estratégia A: Notas de 50 alunos que receberam a Estratégia A.
* Amostra da Estratégia B: Notas de 50 alunos que receberam a Estratégia B.

Usaremos um teste Z para comparar as médias das notas entre as duas amostras.

Se o p-valor do teste Z for menor que um nível de significância pré-determinado (por exemplo, α = 0.05), rejeitamos a hipótese nula e concluímos que há uma diferença significativa nas médias das notas entre as duas estratégias de ensino.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import norm
```

Os dados são criados a seguir:

```python
# Definindo médias e desvios padrão para as notas nas duas estratégias
media_estrategia_A = 70
desvio_padrao_estrategia_A = 10

media_estrategia_B = 75
desvio_padrao_estrategia_B = 12

# Gerando as amostras de notas para cada estratégia de ensino da nossa base
np.random.seed(0)  # Para garantir a reprodutibilidade dos resultados
amostra_estrategia_A = np.random.normal(loc=media_estrategia_A, scale=desvio_padrao_estrategia_A, size=50)
amostra_estrategia_B = np.random.normal(loc=media_estrategia_B, scale=desvio_padrao_estrategia_B, size=50)

print("Notas da Estratégia A:", amostra_estrategia_A[:5])
print("Notas da Estratégia B:", amostra_estrategia_B[:5])
```

# 2) Calcule as médias para as duas amostragens e as variâncias. Quais insights você pode retirar comparando os dados?

# 3) Imprima os resultados da estatística do teste Z, p value e indique se rejeitaremos ou não a hipótese nula. Justifique sua resposta.

```python
from scipy.stats import norm
import numpy as np

# Dados calculados nas questões anteriores
media_A = np.mean(amostra_estrategia_A)
media_B = np.mean(amostra_estrategia_B)

variancia_A = np.var(amostra_estrategia_A, ddof=1)
variancia_B = np.var(amostra_estrategia_B, ddof=1)

n_A = len(amostra_estrategia_A)
n_B = len(amostra_estrategia_B)

alpha = 0.05

# Estatística de teste Z
erro_padrao = np.sqrt(variancia_A / n_A + variancia_B / n_B)
z = (media_B - media_A) / erro_padrao

# p-valor unilateral à direita
p_valor = 1 - norm.cdf(z)

# Resultados
print("=" * 50)
print("TESTE Z PARA DIFERENÇA DE MÉDIAS")
print("=" * 50)

print(f"\nMédia da Estratégia A: {media_A:.2f}")
print(f"Média da Estratégia B: {media_B:.2f}")

print(f"\nVariância da Estratégia A: {variancia_A:.2f}")
print(f"Variância da Estratégia B: {variancia_B:.2f}")

print(f"\nTamanho da amostra A: {n_A}")
print(f"Tamanho da amostra B: {n_B}")

print(f"\nErro padrão da diferença: {erro_padrao:.4f}")

print("\n" + "=" * 50)
print(f"ESTATÍSTICA DE TESTE Z: {z:.4f}")
print(f"P-VALOR (unilateral à direita): {p_valor:.4f}")
print("=" * 50)

print(f"\nNível de significância (alpha): {alpha}")

if p_valor <= alpha:
    print("\nDecisão: Rejeitamos a hipótese nula (H0).")
    print("Conclusão: Há evidência estatística de que a média da Estratégia B é maior que a média da Estratégia A.")
else:
    print("\nDecisão: Não rejeitamos a hipótese nula (H0).")
    print("Conclusão: Não há evidência estatística suficiente, ao nível de 5%, para afirmar que a média da Estratégia B é maior que a média da Estratégia A.")
```

A estatística de teste Z foi de aproximadamente 1,52 e o p-valor foi de aproximadamente 0,064. Como o p-valor é maior que o nível de significância de 0,05, não rejeitamos a hipótese nula. Portanto, não há evidência estatística suficiente para afirmar que a média das notas da Estratégia B é maior do que a média da Estratégia A.

# 4) Para finalizar monte o gráfico da distribuição da estatística do nosso teste Z e explique o que pode ser observado através dele.

```python
# Valores do teste
z_observado = 1.5267
alpha = 0.05

# Valor crítico para teste unilateral à direita
z_critico = norm.ppf(1 - alpha)

# Curva normal padrão
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, color="black", label="Distribuição Normal Padrão")

# Região de rejeição (cauda direita)
x_rejeicao = np.linspace(z_critico, 4, 200)
y_rejeicao = norm.pdf(x_rejeicao)
plt.fill_between(x_rejeicao, y_rejeicao, color="red", alpha=0.4,
                 label=f"Região de rejeição (Z > {z_critico:.2f})")

# Linha do Z observado
plt.axvline(z_observado, color="blue", linestyle="--", linewidth=2,
            label=f"Z observado = {z_observado:.2f}")

# Linha do valor crítico
plt.axvline(z_critico, color="red", linestyle="--", linewidth=2,
            label=f"Z crítico = {z_critico:.2f}")

plt.title("Distribuição da Estatística de Teste Z")
plt.xlabel("Valor de Z")
plt.ylabel("Densidade de probabilidade")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

O gráfico mostra a distribuição normal padrão da estatística de teste Z. A região vermelha, localizada à direita do valor crítico de aproximadamente 1,645, representa a área de rejeição da hipótese nula para um teste unilateral à direita com nível de significância de 5%. O valor observado de Z foi aproximadamente 1,53, ficando à esquerda do valor crítico e fora da região de rejeição. Isso confirma que o p-valor é maior que 0,05 e que não devemos rejeitar a hipótese nula. Portanto, não há evidência estatística suficiente para afirmar que a média das notas da Estratégia B é maior do que a da Estratégia A.
